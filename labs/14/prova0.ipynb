```python
import npfl139
npfl139.require_version("2526.14.1")

task = npfl139.llm.Task.from_name("arithmetic")(seed=42)

data = task.create_dataset(5)

for i, example in enumerate(data):
    print("Example", i)
    print("Prompt:", example.prompt)
    print("Answer:", example.answer)
    print()
```

```python
responses = [
    "The answer is 5.",
    "5",
    "Answer: 5",
    "I think it is five.",
    "The result is -14."
]

for r in responses:
    print(repr(r), "->", task.extract_answer(r))
```

```python
dev = task.create_dataset(5)

# fake_responses = [str(example.answer) for example in dev]
fake_responses = [f"Answer: {example.answer}" for example in dev]

accuracy = task.evaluate(fake_responses, dev)
print("Accuracy:", accuracy)
```
